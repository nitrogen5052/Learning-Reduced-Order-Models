# Introductory LROM figure-design variants

This companion notebook records the figure alternatives considered after the
first paper-layout review.  It preserves the original executed notebook and
changes only the presentation strategy.

The common scientific setup is unchanged: the real Woods--Saxon depth and
radius vary, diffuseness is fixed, the centered wavefunction basis has four
modes, and the learned equation uses two MAXVOL potential evaluations.  FOM,
basis, predictor selection, and LROM calculations come from the repository's `lrom` package.

The revision tests two consistent encodings across both paper figures:

- **color by depth**, with radius distinguished by curve groups or line style;
- **color by radius**, with depth distinguished by curve groups or line style.

The palette is a truncated form of Matplotlib's perceptually ordered
`inferno`, replacing the rainbow-like `turbo` palette used in the first draft
and excluding its nearly black and nearly white endpoints.

White diamonds mark the MAXVOL predictor locations on the central potential.

The four reduced modes use a separate qualitative palette because the basis
belongs to the full manifold rather than to individual parameter values.  The
centered snapshot spectrum is retained as an inset in the basis panel.  One
vertical parameter color bar spans the full composition.

## The Galerkin baseline

The dashed curves below are not a fitted proxy.  They are obtained by
projecting the known scaled radial equation

$$
\phi''(s)+\left[1+\frac{V_0}{E_{\rm cm}}
f_{\rm WS}(s/k;R,a)\right]\phi(s)=0
$$

onto the same centered wavefunction basis used by LS and LROM.  Thus all three
coordinate vectors share one basis and coefficient convention:

- open circles: least-squares projection of the FOM wavefunction;
- solid curves: learned implicit LROM equation;
- dashed curves: conventional Galerkin projection.

Gray regions are outside the training interval of the horizontal parameter.

## Figure 2A: dominant and challenging coordinates

The upper panels retain only $a_1$ and $a_4$.  The narrow lower panels show
absolute coordinate error on a logarithmic scale, so the visual overlap of
LS and LROM does not conceal their difference from Galerkin.  The exact
central reference is retained in the coordinate panels but omitted from the
error panels because its zero centered-coordinate error is tautological; log
limits are set from the remaining noncentral errors.

## Initial visual reading

1. The `inferno` palette is calmer and remains ordered when printed in
   grayscale.
2. Coloring by depth produces the clearest bundle separation when radius is
   the horizontal variable.  Coloring by radius is scientifically equivalent
   and makes the horizontal depth response particularly simple.
3. The $a_1/a_4$ layout communicates the methodological hierarchy most
   directly: LROM lies on the LS coordinates while Galerkin is visibly in a
   different error class.
4. The four-coordinate layout is useful if completeness is more important
   than compactness, but its height is harder to justify in a short paper.
5. No polynomial baseline or two-dimensional error surface appears in these
   main-text candidates; both remain available as supplementary diagnostics in
   notebook 01.
